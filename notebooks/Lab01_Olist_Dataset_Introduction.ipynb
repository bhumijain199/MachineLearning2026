```python
def summarize_table(name, df):
    return {
        "table_name":name,
        "rows":df.shape[0],
        "columns":df.shape[1],
        "missing_values":df.isnull().sum().sum(),
        "duplicate_rows":df.duplicated().sum()
    }
summary = pd.DataFrame([
    summarize_table(name,df)
    for name, df in tables.items()
])
summary
```

```python
summary.to_csv("../reports/lab01_dataset_summary.csv", index=False)
```

```python
def missing_value_summary(df):
    missing_count = df.isnull().sum()
    missing_percent = (missing_count/len(df))*100
    result=pd.DataFrame({
        "missing_count":missing_count,
        "missing_percent":missing_percent
    })
    return result[result["missing_count"]>0].sort_values(
        by="missing_percent",
        ascending=False
    )
missing_value_summary(orders)
```

#Reflection Questions
Q1. Why is it important to understand the dataset before building a Machine Learning model?
Answer
Understanding the dataset before building a Machine Learning model helps identify missing values, duplicate records, data types, and relationships among tables. It also provides business context, enabling better feature selection, data cleaning, and model development. A well-understood dataset leads to more accurate and reliable models.

Q2. Which table appears to be the central table in the Olist dataset? Why?
Answer
The orders table is the central table because it connects customers, order items, payments, and reviews through the order_id and customer_id columns. Most business analyses and Machine Learning tasks revolve around orders, making it the core table of the dataset.

Q3. What is the difference between customer_id and customer_unique_id?
Answer
customer_id uniquely identifies a customer for a particular order.
customer_unique_id uniquely identifies the actual customer across multiple orders.
A single customer may have multiple customer_id values but only one customer_unique_id.
